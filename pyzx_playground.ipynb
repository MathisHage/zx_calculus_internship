# ZX-Calculus Playground

We start by importing PyZX, our main tool.

```python
import pyzx as zx
```

We can generate random Clifford + T and Clifford circuits and visualize them :

```python

qubit_amount = 5
gate_count = 80
#Generate random circuit of Clifford gates
cliffordT = zx.generate.cliffordT(qubit_amount, gate_count)
clifford = zx.generate.cliffords(qubit_amount, gate_count)


#If running in Jupyter, draw the circuit
print("Clifford+T circuit:")
zx.draw(cliffordT)
print("Clifford circuit:")
zx.draw(clifford)
```

We can now simplify the diagrams basically using the pivoting rule :

```python
#Use one of the built-in rewriting strategies to simplify the circuit
zx.simplify.full_reduce(cliffordT)
zx.simplify.full_reduce(clifford)
#See the result
print("Clifford+T circuit after simplification:")
zx.draw(cliffordT)
print("Clifford circuit after simplification:")
zx.draw(clifford)
```

```python
finalT = zx.optimize.basic_optimization(zx.extract_circuit(cliffordT.copy()).to_basic_gates())
zx.draw(finalT)

final = zx.optimize.basic_optimization(zx.extract_circuit(clifford.copy()).to_basic_gates())
zx.draw(final)
```
